# Migrating From C++ to Python


## Basic differences



- Python does not require semicolons, they only serve the purpose of combining two lines into one.
- Python is interpreted, not compiled
    - Compiled languages have a program (the compiler) that turns you source code into computer-readable machine code that you can then execute
    - Interpreted languages have a program (the interpreter) that reads and runs your source code directly (think of how you've made command line programs that've taken inputs before, an interpreter is the exact same thing, but it's reading a coding language)
    - If you have the Python interpreter installed, you can directly run `python myprogram.py` or `python3 myprogram.py` depending on your installation name
    - There is also a Python REPL (read-eval-print-loop) that runs when you just type `python`/`python3`, it reads your input immediately and spits out the output, try opening it and typing `2 + 2`
- Python does not have a main entry point, your entire file can just be `print("Hello World")` and it will run without including iostream or writing the main function.
- Instead of curly brackets `{}` for scoping, Python uses indentation, specifics later.
- Python uses `#` instead of `//` for comments, there is no dedicated syntax for multiline comments like `/* */` in Python, though you can use a string literal like this to mimic it.

````python
"""
This
is
a
multiline
comment
"""
````

## Variable declaration and assignment



C++ requires the type and distinguishes declaration and assignment:
```c++
// declaration
int num;

// assignment
num = 3;

// declaration + assignment
double decimal = 13.4;
std::string word = "12";
```

Python implicitly determines the type and does not distinguish declaration and assignment (this also means declaration is not allowed without assignment alongside it):

````python
# declaration + assignment
num = 3
decimal = 13.4
word = "12"

# assignment
num = 12
````

## if-statements and while loops



 - You will remember that C++ and many other languages define the body of if-statements, while-loops, for-loops, and functions within `{}`
 - Python instead opens these bodies with a `:` instead of a `{` and instead of having `}` to close the body, the first line of non-indented text is the end of the body
 - Python also does not require () around the conditional, and else if is shortened to elif

```c++
if (false) {
    printf("Wow I didn't reach here\n");
}
else if (true) {
    printf("I will probably reach here\n");
}
else {
    printf("I should not be reaching here either\n");
}
```

````python
if False:
    print("Wow I didn't reach here")
elif True:
    print('I will probably reach here')
else:
    print('I should not be reaching here either')
````

````python
i = 0
max = 5
while i < max:
    # this is a nifty format specifier that automatically prints the variable name with its value
    print(f'{i=}')
    # Python removed ++ and -- for some reason
    i += 1
    
````

## Lists and Tuples

- Lists are the equivelant of std::vector; they're dynamic arrays
- Tuples similarily are containers of multiple values but are completely immutable
- You still use [] to index both of these, with indices starting at 0 and ending at length - 1
- Lists use [x, y, ...] for the literal syntax while tuples use (x, y, ...) for the literal syntax
- Other than the mutability and literal syntax differences, lists and tuples are mostly the same

````python
lst = [1, 2, 3, 4, 5]
print(f'{lst[2]=}')

# gets the size in # of items of the list/tuple similar to std::vector.size / std::array.size()
print(f'{len(lst)=}')

lst[4] = 12
print(f'{lst=}')

# adds to the end of the list
lst.append(9)
print(f'{lst=}')

# you can even have different data types in lists and tuples 
# (though you will only see this in tuples since you'll know which indices correspond to which datatype)
lst.append('stringggggg')
print(f'{lst=}')
````

````python
tup = (1, 2, 3, 4, 5)
print(f'{tup[2]=}')

print(f'{len(tup)=}')

# cannot modify tuples so this will give an error
tup[4] = 12
print(f'{tup=}')
````

## For-loops introduction


 - Python for-loops are quite unique
    - Unlike in languages like C++ or Javascript, Python for-loops are not a concise way to write the most common while-loop pattern: a repeater
    - Python for-loops are more like foreach-loops
    - Recall in C++ (`list` is a `std::vector<int>`):
    ```c++
    for (int num : list) {
      std::cout << num  << '\n';
    }
    ```

    - In Python, the equivelant is this:

````python
lst = [1, 2, 3, 4, 5]
for num in lst:
    print(f'{num=}')
````

What if you want the tradtional for-loop as a repeater?
 ```c++
 for (int i = 0; i < 5; i++) {
    std::cout << i << '\n';
 }
 ```
Since Python for-loops are for-each loops, you have to somehow mold the classic repeater pattern into an iterable, done via:

````python
for i in range(5):
    print(f'{i=}')
````

If you still want to keep your index while iterating through your list, you can use enumerate

````python
lst = [9, 6, 7, 5, 2, 4]
for i, num in enumerate(lst):
    print(f'{i=}, {num=}')
````

## Functions

You do not have to specify the return type of a function specify the data types of the parameters

````python
def func(param1, param2, param3):
    print(param1)
    return param2 + param3

# function calls are exactly what you would expect them to be
return_value = func('doggy', 12, 14)
print(f'{return_value=}')
````

Functions specifically are finnicky when it comes to global variables

````python
# Globally defined PI
PI = 3.14159

def weird_get_area(radius):
    # This PI is a local variable PI, it has nothing to do with the globally defined PI
    PI = 4
    return PI * radius * radius

print(f'{weird_get_area(5)=}')

# You'll see here that even though we modified PI in the get_area function, we only modified the local one
# The global one is unaffected as evident here
print(f'{PI=}')
````

````python
# Globally defined PI
PI = 3.14159

def weird_get_area(radius):
    # The global keyword allows the function to access the global variable PI instead of making a local one
    global PI
    PI = 4
    return PI * radius * radius

print(f'{weird_get_area(5)=}')

# The global one is now affected
print(f'{PI=}')
````
